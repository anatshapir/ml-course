````python
#@title הכנה - לחצו להריץ
import os, subprocess, sys
try:
    from bidi.algorithm import get_display
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "python-bidi"], check=True)
    from bidi.algorithm import get_display

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def heb(text):
    """מחזיר טקסט עברי שמוצג נכון בגרפים (matplotlib לא הופך כיוון בעצמו)."""
    return get_display(text)

def check(test, ok="נכון!", hint="עוד לא."):
    """בדיקת תרגיל: מריצה את test ומדפיסה משוב, בלי לעצור את המחברת."""
    try:
        passed = bool(test())
    except Exception:
        passed = False
    from IPython.display import HTML, display
    style = ("background:#DCFCE7;color:#14532D;border-right:6px solid #16A34A" if passed
             else "background:#FEE2E2;color:#7F1D1D;border-right:6px solid #DC2626")
    msg = ("✅ " + ok) if passed else ("❌ " + hint)
    display(HTML(f'<div dir="rtl" style="{style};padding:10px 14px;border-radius:8px;'
                 f'font-family:Arial;font-size:17px;margin:4px 0">{msg}</div>'))

plt.rcParams["figure.figsize"] = (8, 4)
RANDOM_STATE = 42

DATA_BASE = "https://raw.githubusercontent.com/anatshapir/datasets/main/"

def load_data(name):
    """טוען קובץ נתונים של הקורס: מתיקייה מקומית data/ אם יש, אחרת ממאגר הקורס ב-GitHub."""
    local = os.path.join("data", name)
    return pd.read_csv(local if os.path.exists(local) else DATA_BASE + name, encoding="utf-8-sig")

df = load_data("towns_turnout_19_25.csv")
print("הנתונים נטענו:", df.shape[0], "שורות,", df.shape[1], "עמודות")

# ---- בדיקות התרגילים (נוצר אוטומטית) ----
_CHECKS = {
    "תרגיל 1": (lambda: town == "דימונה" and abs(change - 10.55) < 0.001 and went_up is True,
        "נכון! השינוי בדימונה: 10.55 נקודות.", "town במירכאות, change = t25 - t19, ו-went_up היא השוואה שנותנת True או False"),
    "תרגיל 2": (lambda: first_town == my_towns[0] and last_change == my_changes[-1] and n == len(my_towns)
      and middle == my_towns[2:5] and has_eilat == ("אילת" in my_towns),
        "נכון!", "אינדקס ראשון 0, אחרון מינוס 1, וחיתוך לא כולל את המקום שבסוף"),
    "תרגיל 3": (lambda: n10 == len(my_changes) and biggest_rise == max(my_changes)
      and avg_change == round(sum(my_changes) / len(my_changes), 2),
        "נכון!", "ממוצע = סכום חלקי מספר הערכים, ו-round מסביב לכל החישוב"),
    "תרגיל 4": (lambda: worst_change == min(my_changes) and worst_i == my_changes.index(min(my_changes))
      and worst_town == my_towns[my_changes.index(min(my_changes))],
        "נכון! ערך, מקום, שם.", "min נותן ערך; my_changes.index(ערך) נותן מקום; my_towns[מקום] נותן שם"),
    "תרגיל 5": (lambda: worst_change_all == min(changes) and worst_town_all == names[changes.index(min(changes))]
      and best_change_all == max(changes) and best_town_all == names[changes.index(max(changes))],
        "תרגיל 5 נכון.", "תרגיל 5: אותם שלושה צעדים כמו בתרגיל 4, על names ו-changes"),
    "תרגיל 6": (lambda: worst_voters == voters[changes.index(min(changes))] and best_voters == voters[changes.index(max(changes))]
      and total_voters == sum(voters),
        "תרגיל 6 נכון.", "תרגיל 6: אותו מקום, ברשימה voters"),
    "תרגיל 7": (lambda: avg_all == round(sum(changes) / len(changes), 2) and first == names[0],
        "נכון! מצאתם גם את השגיאות שאין להן הודעה.", "יש 1177 יישובים, לא 1000. והמקום הראשון ברשימה הוא 0"),
    "תרגיל 8": (lambda: top3 == sorted(changes)[:3] and top3_towns == [names[changes.index(x)] for x in sorted(changes)[:3]],
        "נכון!", "קודם sorted וחיתוך; אחר כך שלוש פעמים names[changes.index(...)]"),
    "בית 2": (lambda: avg_all_hw == round(sum(changes) / len(changes), 2) and biggest_town == names[voters.index(max(voters))],
        "נכון!", "ממוצע: sum חלקי len. היישוב הגדול: הערך הגדול ב-voters, המקום שלו, והשם במקום הזה"),
    "בית 3": (lambda: j == names.index("ירושלים") and jlm_change == changes[names.index("ירושלים")]
      and jlm_voters == voters[names.index("ירושלים")]
      and times == round(voters[names.index("ירושלים")] / voters[changes.index(min(changes))]),
        "נכון!", ".index עובד גם על רשימת שמות. ואת היישוב עם הירידה הגדולה מצאתם בתרגיל 5"),
    "בית 5": (lambda: middle_value == sorted(changes)[len(changes) // 2],
        "נכון! לערך הזה קוראים חציון - נפגוש אותו בשבועות הסטטיסטיקה.", "קודם ממיינים, אחר כך לוקחים את המקום len(changes) // 2"),
}

def check_ex(key):
    """בודק תרגיל לפי שמו. הבדיקה עצמה מוגדרת למעלה, בתא ההכנה."""
    test, ok, hint = _CHECKS[key]
    check(test, ok=ok, hint=hint)
````

| מושג | ג'אווה | פייתון |
|---|---|---|
| משתנה | `double t = 53.02;` | `t = 53.02` - בלי טיפוס ובלי נקודה-פסיק |
| טיפוסים בסיסיים | `int, double, boolean, char, String` | `int, float, bool, str` - אין `char` |
| חילוק | `7 / 2` נותן 3 | `7 / 2` נותן 3.5, ו-`7 // 2` נותן 3 |
| רשימה | `int[] a = {1, 2};` | `a = [1, 2]` |
| אורך | `a.length` | `len(a)` |
| האיבר האחרון | `a[a.length - 1]` | `a[-1]` |
| חיתוך | `s.substring(1, 3)` | `a[1:3]` - הסוף לא כלול |
| פונקציות מוכנות | - | `min(a)`, `max(a)`, `sum(a)`, `sorted(a)`, `a.index(x)` |

**מה חשוב לזכור**
1. אינדקסים מתחילים ב-0, ו-`a[1:3]` לא כולל את מקום 3.
2. `max(a)` נותן **ערך**. `a.index(ערך)` נותן את **המקום** שלו. ובמקום הזה, ברשימה אחרת, נמצא השם.
3. שגיאה עם הודעה - קוראים את השורה האחרונה. שגיאה לוגית - אין הודעה, ולכן בודקים.
